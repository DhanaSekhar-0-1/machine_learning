**Orthonormal vectors:**
Vectors are orthogonal and each has unit length:∥𝑉∥

we verify whether the given vectors are orthogonal by computing pairwise dot products.
Then, we normalize each vector to obtain orthonormal vectors.

**Algorithm**
Algorithm (Orthogonal Check)

Input vectors
𝑉

𝑉

Compute dot products

V1⋅V2

V1⋅V3

V2⋅V3
	​

.

If all dot products are zero → vectors are orthogonal.

Otherwise → not orthogonal.

```python
import numpy as np

V1 = np.array([3, 5, 4])
V2 = np.array([3, -5, 4])
V3 = np.array([4, 0, -3])

dot12 = np.dot(V1, V2)
dot13 = np.dot(V1, V3)
dot23 = np.dot(V2, V3)

if dot12 == 0 and dot13 == 0 and dot23 == 0:
    print("Vectors are Orthogonal")
else:
    print("Vectors are NOT Orthogonal")

print("Dot Products:", dot12, dot13, dot23)
```

**Algorithm (Orthonormal Conversion)**

1. Compute magnitude of each vector.

2. Divide vector by its magnitude.

3. Output normalized vectors.

```python
def normalize(v):
    return v / np.linalg.norm(v)

V1_n = normalize(V1)
V2_n = normalize(V2)
V3_n = normalize(V3)

print("Orthonormal Vectors:")
print("V1:", V1_n)
print("V2:", V2_n)
print("V3:", V3_n)
```

**Time Complexity (Part i)**

Dot product of vectors of size n: O(n)

Normalization: O(n)
Total Time Complexity: O(n)

**Space Complexity (Part i)**

Constant extra variables
Space Complexity: O(1)

**we need to calculate the angle between each pair of vectors using the cosine formula and convert the result from radians to degrees**


**Algorithm**

1. Compute dot product of vectors.

2. Compute magnitudes.

3. Apply cosine formula.

4. Convert angle to degrees.

```python
import math

def angle_between(A, B):
    cos_theta = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
    return math.degrees(math.acos(cos_theta))

print("Angle between V1 and V2:", angle_between(V1, V2))
print("Angle between V1 and V3:", angle_between(V1, V3))
print("Angle between V2 and V3:", angle_between(V2, V3))
```
