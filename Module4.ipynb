**INVERSE OF IDENTITY MATRIX : (I^-1=I)**

The inverse of an identity matrix is the identity matrix itself.


**Why it matters**
*   Defines matrix inverse
*   Every inverse operation is judged against I:
*   Linear equation solving
*   Orthogonal matrices


**Algorithm**
*   Create identity matrix I.
*   Compute inverse of I.
*   Compare I^−1 with I
*   Use if–else to verify the property.

```python
import numpy as np

# Sample input: Identity matrix
I = np.array([[1, 0, 0],
              [0, 1, 0],
              [0, 0, 1]])

# Find inverse
I_inv = np.linalg.inv(I)

# Verify property using if-else
if np.allclose(I_inv, I):
    print("Property Verified: I^-1 = I")
else:
    print("Property Not Verified")

```

**The Above code Proves that I^-1 is egual to I**

**and It disqualifies the I^-1 is not equal to I**

**Time Complexity**

Transpose: O(n²)

Comparison: O(n²)

Total Time Complexity: O(n²)

**Space Complexity**

No extra memory used

Extra Space Complexity: O(1)
