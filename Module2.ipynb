**Orthogonal Matrix : (A^−1=A^T) or (A^T A=I)**

An orthogonal matrix is a square matrix whose transpose is also its inverse.

No sugar-coating: it’s a matrix that preserves length and angles. Nothing stretches, nothing shrinks.

**properties**

Columns and rows are orthonormal vectors
*   unit length
*   mutually perpendicular

Determinant = +1 or −1
*   +1 → pure rotation
*   −1 → rotation + reflection

Preserves:
*   vector length
*   dot product
*   angles


---


**Why orthogonal matrices matter**
*   Computer graphics → rotations without distortion

*   Machine learning → numerical stability

*   Physics → coordinate transformations
*   Signal processing → energy preservation


**Algorithm**

1.   Read matrix A.

2.   Ensure matrix is square.

3.   Compute  𝐴^𝑇 𝐴

4.   Compare result with identity matrix.

5.   Use if–else to verify orthogonality.

6.   List item



```python
import numpy as np

A = np.array([[0, -1],
              [1,  0]])

I = np.eye(A.shape[0])

if A.shape[0] == A.shape[1] and np.allclose(A.T @ A, I):
    print("Given Matrix is Orthogonal (A^T = A^-1)")
else:
    print("Given Matrix is NOT Orthogonal")
```

**Time Complexity**

Matrix multiplication
𝐴^𝑇 𝐴: O(n³)

Comparison with identity: O(n²)
Total Time Complexity: O(n³)

**Space Complexity**

Identity matrix storage
Extra Space Complexity: O(1)
