**Symmetric Matrix: A^T=A**
  symmetric matrix is a square matrix that is equal to its transpose matrix.

**properties**

*  IT Must be a square (n × n) Matrix .

*   Rectangular matrices can’t be symmetric.

*  Diagonal elements stay the same after transpose.

*  All symmetric matrices are equal across the main diagonal.

* Used heavily in linear algebra, machine learning, physics, and optimization.















**Algorithm**

Read matrix A

Check whether the matrix is square.

Compute transpose
𝐴^𝑇

Compare
𝐴
  and
𝐴^𝑇


Use if–else to verify the given matrix is symmetric matrix or not and then print the result.

```python
import numpy as np

A = np.array([[1, 2, 3],
              [2, 4, 5],
              [3, 5, 6]])

if A.shape[0] == A.shape[1] and np.allclose(A, A.T):
    print("Matrix is Symmetric (A^T = A)")
else:
    print("Matrix is NOT Symmetric")
```

**Time Complexity**

Transpose operation: O(n²)

Comparison: O(n²)
Total Time Complexity: O(n²)

**Space Complexity**

Transpose uses a view, not a copy
Extra Space Complexity: O(1)
