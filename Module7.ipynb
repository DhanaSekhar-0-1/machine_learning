Given Matrix A
=
   [1     x  5]
   
   [2-2i  2  5]

   [5     5  3]



---
	​A matrix is symmetric if:
   A^T=A
=
To make the given matrix symmetric, the elements on either side of the main diagonal must be equal.
By comparing corresponding elements, we can find the value of x.

```python
import numpy as np

# Value of x from symmetry condition
x = 2 - 2j

A = np.array([[1, x, 5],
              [2 - 2j, 2, 5],
              [5, 5, 3]])

if np.allclose(A, A.T):
    print("Matrix is Symmetric")
    print("Value of x =", x)
else:
    print("Matrix is NOT Symmetric")
```

**Time Complexity**

Matrix comparison: O(n²)

Time Complexity: O(n²)

**Space Complexity**

Constant extra space

Space Complexity: O(1)
