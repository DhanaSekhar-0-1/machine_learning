**IDENTITY MATRIX : (I^T=I)**

An identity matrix is a square matrix that acts like the number 1 in matrix multiplication.

Blunt truth: it changes nothing when you multiply by it.

**properties**

*   Must be square
*   A⋅I=I⋅A=A
*   Diagonal elements = 1
*   Non-diagonal elements = 0
*   Determinant = 1
*   Inverse of identity = identity

---

Why it matters

Used to define inverse matrices

Appears in linear equations

Core to orthogonal matrices:

A^T.A=I

Fundamental in computer graphics, ML, physics










**Algorithm**
*   Create identity matrix I.
*   Compute transpose I^T
*   Compare both matrices.
*   Use if–else to verify the property.

```python
import numpy as np

I = np.eye(3)

if np.allclose(I, I.T):
    print("Identity matrix satisfies (I^T = I)")
else:
    print("Property NOT satisfied")
```

**Time Complexity**

Transpose: O(n²)

Comparison: O(n²)

Total Time Complexity: O(n²)

**Space Complexity**

No extra memory used

Extra Space Complexity: O(1)
