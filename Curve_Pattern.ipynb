**Case 1: Underfitting (High Bias)**

**Pattern**

Training error: high

Validation error: high

`Curves are close together`

**Meaning**

Model is too simple. More data won’t help.

**Fix**

Increase model complexity

Add features

Reduce regularization

**alpha = [0.0001, 1, 1000]**


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve
from sklearn.linear_model import Ridge
from sklearn.datasets import make_regression

# Create synthetic data
X, y = make_regression(
    n_samples=500,
    n_features=10,
    noise=20,
    random_state=42
)

# Model (change alpha to see regularization effect)
model = Ridge(alpha=1.0)

# Compute learning curve
train_sizes, train_scores, val_scores = learning_curve(
    model,
    X,
    y,
    cv=5,
    scoring='neg_mean_squared_error',
    train_sizes=np.linspace(0.1, 1.0, 10)
)

# Convert to positive MSE
train_error = -train_scores.mean(axis=1)
val_error = -val_scores.mean(axis=1)

# Plot
plt.figure()
plt.plot(train_sizes, train_error, label="Training Error")
plt.plot(train_sizes, val_error, label="Validation Error")
plt.xlabel("Training Set Size")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.show()
```

**Situation	Training Error	Validation Error	Diagnosis**
**Both high, close	High	High	Underfitting**
**Big gap	Low	High	Overfitting**
**Both low, close	Low	Low	Good fit**
