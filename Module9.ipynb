```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, skew
```

```python
# Load CSV file
file_path = "/data_set.csv"
df = pd.read_csv(file_path)

# Clean and select numeric data (Average stay in hospital)
data = pd.to_numeric(df.iloc[1:, 1], errors="coerce").dropna()
```

```python
# a) Mean, Median, Mode
mean_val = data.mean()
median_val = data.median()
mode_val = data.mode().values

print("Mean   :", round(mean_val, 2))
print("Median :", round(median_val, 2))
print("Mode   :", mode_val)
```

```python
# b) Skewness
skewness_val = skew(data, bias=False)
print("Skewness :", round(skewness_val, 2))
```

```python
# Gaussian Distribution
x = np.linspace(data.min(), data.max(), 100)
gaussian_curve = norm.pdf(x, mean_val, data.std())
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.hist(data, bins=10, density=True)
plt.plot(x, gaussian_curve)
plt.title("Gaussian Distribution of Hospital Stay (AMI, 2020)")
plt.xlabel("Average Length of Stay (Days)")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```
